# Copy HOSIX Source to HOSIX-GEPROSALUD
This notebook copies the minimal HOSIX source and required shared UI/hook files into the new `HOSIX-GEPROSALUD` repo and verifies the destination contents.

```python
from pathlib import Path
source_root = Path(r"c:\Users\HP\Desktop\Proyectos y Empresas\geprostec\RENAPROSA\Renaprosa2\SERMED2")
target_root = Path(r"c:\Users\HP\Desktop\Proyectos y Empresas\geprostec\RENAPROSA\Renaprosa2\HOSIX-GEPROSALUD")
print('Source root:', source_root)
print('Target root:', target_root)
assert source_root.exists(), 'Source root missing'
assert target_root.exists(), 'Target root missing'
print('Source exists:', source_root.exists())
print('Target exists:', target_root.exists())
```

```python
import shutil
paths_to_copy = [
    'src/components/hosix',
    'src/pages/Hosix',
    'src/components/ui',
    'src/hooks/useHosixAuth.ts',
    'src/integrations/supabase/hosixClient.ts',
    'src/integrations/supabase/types.ts',
    'src/index.css',
    'package.json',
    'vite.config.ts',
    'tsconfig.json',
    'tsconfig.node.json',
    'postcss.config.js',
    'tailwind.config.ts',
    'index.html',
    '.gitignore',
]

copied = []
missing = []
for rel in paths_to_copy:
    src = source_root / rel
    dst = target_root / rel
    print('Copying', src, '->', dst)
    if not src.exists():
        print('  MISSING:', src)
        missing.append(str(src))
        continue
    if src.is_dir():
        shutil.copytree(src, dst, dirs_exist_ok=True)
    else:
        dst.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy2(src, dst)
    copied.append(str(src))

print('\nCopied:', len(copied), 'paths')
print('Missing:', len(missing), 'paths')
for path in missing:
    print('  ', path)

print('\nTarget structure:')
for p in sorted(target_root.glob('**/*')):
    print(' ', p.relative_to(target_root))
```
